```python
import numpy as np

# Define the perceptron class
class Perceptron:
    def __init__(self, input_size, learning_rate=0.1, epochs=10):
        self.weights = np.zeros(input_size + 1)  # +1 for the bias
        self.learning_rate = learning_rate
        self.epochs = epochs

    def predict(self, x):
        # Linear combination of inputs and weights
        linear_combination = np.dot(x, self.weights[1:]) + self.weights[0]
        # Apply step function
        return 1 if linear_combination > 0 else 0

    def train(self, training_data, labels):
        for _ in range(self.epochs):
            for x, y in zip(training_data, labels):
                prediction = self.predict(x)
                # Update weights and bias if there's a misclassification
                if prediction != y:
                    error = y - prediction
                    self.weights[1:] += self.learning_rate * error * x
                    self.weights[0] += self.learning_rate * error

# Training data for AND gate
and_training_data = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
and_labels = np.array([0, 0, 0, 1])

# Training data for OR gate
or_training_data = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
or_labels = np.array([0, 1, 1, 1])

# Train perceptron for AND gate
and_perceptron = Perceptron(input_size=2)
and_perceptron.train(and_training_data, and_labels)

# Train perceptron for OR gate
or_perceptron = Perceptron(input_size=2)
or_perceptron.train(or_training_data, or_labels)

# Test the trained perceptrons on AND gate inputs
and_results = [and_perceptron.predict(x) for x in and_training_data]

# Test the trained perceptrons on OR gate inputs
or_results = [or_perceptron.predict(x) for x in or_training_data]

print("AND gate results:", and_results)
print("OR gate results:", or_results)
```
