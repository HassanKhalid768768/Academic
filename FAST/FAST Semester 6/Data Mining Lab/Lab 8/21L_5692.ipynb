```python
import pandas as pd
import numpy as np
from apyori import apriori
```

```python
df = pd.read_csv('/content/groceries_dataset.csv')
```

```python
records = []
for i in range(0, 9835):
    records.append([str(df.values[i,u]) for u in range(0, 32)])
```

```python
rules = apriori(records, min_support=0.0022, min_confidence=0.20, min_lift=3, min_length = 4)
```

```python
# Store rules in result variable
results = list(rules)# See the items that were bought together with their support
results_list = []
for i in range(0, len(results)):
    results_list.append('RULE:\t' + str(results[i][0]) + '\nSUPPORT:\t' + str(results[i][1]))
```

```python
columns=["age","workclass","fnlwgt","education","education-num","marital-status","occupation","relationship","race","sex","capital-gain","capital-loss","hours-per-week","native-country","income"]
df = pd.read_csv('/content/DATA.csv',header=None,names=columns)
df.head()
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64', 'int32']).columns
num_cols
```

```python
# One-Hot Encoding
from sklearn.preprocessing import OneHotEncoder

categorical_cols = ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'native-country']

# Create a new DataFrame with only relevant columns
categorical_df = df[categorical_cols]
num_df=df[num_cols]
encoder = OneHotEncoder()

# Fit and transform the categorical data
encoded_data = encoder.fit_transform(categorical_df).toarray()

# Create a DataFrame from the encoded data
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(categorical_cols))

# Combine the encoded DataFrame with the original numeric columns
binarized_df = pd.concat([encoded_df], axis=1)

# Display the combined DataFrame
print(binarized_df.head())
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

# Set your support and confidence thresholds
min_support_values = [0.05, 0.1, 0.15]
min_confidence = 0.7

# Calculate frequent itemsets
for min_support in min_support_values:
    frequent_itemsets = apriori(binarized_df, min_support=min_support, use_colnames=True)

    # Extract association rules
    rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=min_confidence)

    # Rank the top 10 rules
    top_10_rules = rules.nlargest(10, "confidence")

    # Print rules and explanations
    print(f"Rules for min_support={min_support:.2f}:")
    for i, (_, row) in enumerate(top_10_rules.iterrows(), start=1):
        antecedent = row['antecedents']
        consequent = row['consequents']
        confidence = row['confidence']

        print(f"Rule {i}: If {antecedent} then {consequent} (Confidence: {confidence:.2f})")
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

def calculate_interest(rule):
    return rule['confidence'] * rule['lift']

min_support_values = [0.05, 0.1, 0.15]
min_interest_values = [0.5, 0.7, 0.9]

for min_support in min_support_values:
    frequent_itemsets = apriori(binarized_df, min_support=min_support, use_colnames=True)

    for min_interest in min_interest_values:
        rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=min_confidence)

        # Calculate interest measure for each rule
        rules['interest'] = rules.apply(calculate_interest, axis=1)

        # Rank the top 10 rules based on interest
        top_10_rules = rules.nlargest(10, "interest")

        # Print rules and explanations
        print(f"Rules for min_support={min_support:.2f} and min_interest={min_interest:.2f}:")
        for i, (_, row) in enumerate(top_10_rules.iterrows(), start=1):
            antecedent = row['antecedents']
            consequent = row['consequents']
            confidence = row['confidence']
            lift = row['lift']
            interest = row['interest']

            print(f"Rule {i}: If {antecedent} then {consequent} (Confidence: {confidence:.2f}, Lift: {lift:.2f}, Interest: {interest:.2f})")
```

```python
# c) Formulate questions and select attributes for association rule mining
# For example, let's say we want to find out rules related to high-income individuals who work in managerial positions and have a higher education level.

selected_attributes = ['education', 'occupation', 'income']

# Select relevant columns from the DataFrame
selected_df = df[selected_attributes]

# Binarize the selected DataFrame
selected_encoded_data = encoder.fit_transform(selected_df).toarray()
selected_encoded_df = pd.DataFrame(selected_encoded_data, columns=encoder.get_feature_names_out(selected_attributes))

# Calculate frequent itemsets and association rules
frequent_itemsets_selected = apriori(selected_encoded_df, min_support=0.1, use_colnames=True)
rules_selected = association_rules(frequent_itemsets_selected, metric="confidence", min_threshold=0.7)

# Display the top 10 rules
print("Top 10 rules based on selected attributes:")
for i, (_, row) in enumerate(rules_selected.nlargest(10, "confidence").iterrows(), start=1):
    antecedent = row['antecedents']
    consequent = row['consequents']
    confidence = row['confidence']
    lift = row['lift']
    print(f"Rule {i}: If {antecedent} then {consequent} (Confidence: {confidence:.2f}, Lift: {lift:.2f})")
```

2.   If a person's highest education level is high school graduation (HS-grad), then they are likely to have an income of less than or equal to $50K (income_ <=50K) with a confidence of 0.84 and lift of 1.11.

These rules provide insights into the relationship between occupation and education level with income status. They suggest that individuals in administrative or clerical positions or with lower education levels (such as high school graduates or those with some college education) are more likely to have incomes below $50K. This insight could be useful for various purposes, such as targeted marketing strategies, understanding socioeconomic demographics, or designing educational and career development programs.

```python
# d) Try evaluation measures other than confidence and lift
# For example, we can try using leverage as an evaluation measure

rules_leverage = association_rules(frequent_itemsets_selected, metric="leverage", min_threshold=0.0)

# Display the top 10 rules based on leverage
print("Top 10 rules based on leverage:")
for i, (_, row) in enumerate(rules_leverage.nlargest(10, "leverage").iterrows(), start=1):
    antecedent = row['antecedents']
    consequent = row['consequents']
    confidence = row['confidence']
    leverage = row['leverage']
    print(f"Rule {i}: If {antecedent} then {consequent} (Confidence: {confidence:.2f}, Leverage: {leverage:.2f})")
```
