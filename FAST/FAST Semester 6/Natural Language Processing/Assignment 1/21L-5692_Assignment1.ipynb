```python
import os.path
import sys
import random
import math
from collections import defaultdict

#----------------------------------------
#  Data input 
#----------------------------------------

# Read a text file into a corpus (list of sentences (which in turn are lists of words))
def readFileToCorpus(f):
    """ Reads in the text file f which contains one sentence per line.
    """
    if os.path.isfile(f):
        file = open(f, "r")
        i = 0
        corpus = []
        print("Reading file ", f)
        for line in file:
            i += 1
            sentence = line.split()
            corpus.append(sentence)
            if i % 1000 == 0:
                sys.stderr.write("Reading sentence " + str(i) + "\n")
        return corpus
    else:
        print("Error: corpus file ", f, " does not exist")
        sys.exit()

# Preprocess the corpus
def preprocess(corpus):
    # Find all the rare words
    freqDict = defaultdict(int)
    for sen in corpus:
        for word in sen:
            freqDict[word] += 1

    # Replace rare words with UNK
    for sen in corpus:
        for i in range(len(sen)):
            word = sen[i]
            if freqDict[word] < 2:
                sen[i] = UNK

    # Bookend the sentences with start and end tokens
    for sen in corpus:
        sen.insert(0, start)
        sen.append(end)
    
    return corpus

def preprocessTest(vocab, corpus):
    # Replace test words that were unseen in the training with UNK
    for sen in corpus:
        for i in range(len(sen)):
            word = sen[i]
            if word not in vocab:
                sen[i] = UNK
    
    # Bookend the sentences with start and end tokens
    for sen in corpus:
        sen.insert(0, start)
        sen.append(end)

    return corpus

# Constants 
UNK = "UNK"     # Unknown word token
start = "<s>"   # Start-of-sentence token
end = "</s>"    # End-of-sentence token


#--------------------------------------------------------------
# Language models and data structures
#--------------------------------------------------------------

# Parent class for the three language models
class LanguageModel:
    def __init__(self, corpus):
        pass

    def generateSentence(self):
        print("Implement the generateSentence method in each subclass")
        return []

    def getSentenceProbability(self, sen):
        print("Implement the getSentenceProbability method in each subclass")
        return 0.0

    def getCorpusPerplexity(self, corpus):
        print("Implement the getCorpusPerplexity method")
        return 0.0

    def generateSentencesToFile(self, numberOfSentences, filename):
        filePointer = open(filename, 'w+')
        for i in range(numberOfSentences):
            sen = self.generateSentence()
            prob = self.getSentenceProbability(sen)

            stringGenerated = str(prob) + " " + " ".join(sen) 
            print(stringGenerated, end="\n", file=filePointer)
        filePointer.close()

# Unigram language model
class UnigramModel(LanguageModel):
    def __init__(self, corpus):
        self.unigram_dist = UnigramDist(corpus)

    def generateSentence(self):
        sentence = [start]
        while True:
            word = self.unigram_dist.draw()
            if word == end:
                break
            sentence.append(word)
        return sentence

    def getSentenceProbability(self, sen):
        prob = 1.0
        for word in sen[1:-1]:  
            prob *= self.unigram_dist.prob(word)
        return prob

    def getCorpusPerplexity(self, corpus):
        log_prob_sum = 0.0
        total_words = 0
        for sen in corpus:
            total_words += len(sen) - 1  
            for word in sen[1:-1]: 
                log_prob_sum += math.log(self.unigram_dist.prob(word))
        avg_log_prob = log_prob_sum / total_words
        perplexity = math.exp(-avg_log_prob)
        return perplexity

# Smoothed unigram language model (use Laplace smoothing)
class SmoothedUnigramModel(LanguageModel):
    def __init__(self, corpus):
        self.unigram_dist = UnigramDist(corpus)

    def generateSentence(self):
        sentence = [start]
        while True:
            word = self.unigram_dist.draw()
            if word == end:
                break
            sentence.append(word)
        return sentence

    def getSentenceProbability(self, sen):
        prob = 1.0
        V = len(self.unigram_dist.counts)  
        N = self.unigram_dist.total  
        for word in sen[1:-1]:  
            prob *= (self.unigram_dist.counts[word] + 1) / (N + V)
        return prob

    def getCorpusPerplexity(self, corpus):
        log_prob_sum = 0.0
        total_words = 0
        V = len(self.unigram_dist.counts)  
        N = self.unigram_dist.total  
        for sen in corpus:
            total_words += len(sen) - 1  
            for word in sen[1:-1]:  
                log_prob_sum += math.log((self.unigram_dist.counts[word] + 1) / (N + V))
        avg_log_prob = log_prob_sum / total_words
        perplexity = math.exp(-avg_log_prob)
        return perplexity


# Unsmoothed bigram language model
class BigramModel(LanguageModel):
    def __init__(self, corpus):
        self.bigram_counts = defaultdict(lambda: defaultdict(int))
        self.unigram_counts = defaultdict(int)
        self.train(corpus)

    def train(self, corpus):
        for sen in corpus:
            for i in range(len(sen) - 1):
                self.bigram_counts[sen[i]][sen[i+1]] += 1
                self.unigram_counts[sen[i]] += 1

    def generateSentence(self):
        sentence = [start]
        while True:
            prev_word = sentence[-1]
            word = random.choices(list(self.bigram_counts[prev_word].keys()), 
                                  list(self.bigram_counts[prev_word].values()))[0]
            if word == end:
                break
            sentence.append(word)
        return sentence

    def getSentenceProbability(self, sen):
        prob = 1.0
        for i in range(len(sen) - 1):
            prev_word = sen[i]
            word = sen[i+1]
            bigram_count = self.bigram_counts[prev_word][word]
            unigram_count = self.unigram_counts[prev_word]
            bigram_prob = (bigram_count + 1) / (unigram_count + len(self.bigram_counts[prev_word]))
            prob *= bigram_prob
        return prob

    def getCorpusPerplexity(self, corpus):
        log_prob_sum = 0.0
        total_words = 0
        for sen in corpus:
            total_words += len(sen) - 1  
            for i in range(len(sen) - 1):
                prev_word = sen[i]
                word = sen[i+1]
                bigram_count = self.bigram_counts[prev_word][word]
                unigram_count = self.unigram_counts[prev_word]
                bigram_prob = (bigram_count + 1) / (unigram_count + len(self.bigram_counts[prev_word]))
                log_prob_sum += math.log(bigram_prob)
        avg_log_prob = log_prob_sum / total_words
        perplexity = math.exp(-avg_log_prob)
        return perplexity

# Smoothed bigram language model (use linear interpolation smoothing, set lambda1 = lambda2 = 0.5)
class SmoothedBigramModelInterpolation(LanguageModel):
    def __init__(self, corpus):
        self.bigram_counts = defaultdict(lambda: defaultdict(int))
        self.unigram_counts = defaultdict(int)
        self.unigram_dist = UnigramDist(corpus)
        self.train(corpus)

    def train(self, corpus):
        for sen in corpus:
            for i in range(len(sen) - 1):
                self.bigram_counts[sen[i]][sen[i+1]] += 1
                self.unigram_counts[sen[i]] += 1

    def generateSentence(self):
        sentence = [start]
        while True:
            prev_word = sentence[-1]
            word = random.choices(list(self.bigram_counts[prev_word].keys()), 
                                  list(self.bigram_counts[prev_word].values()))[0]
            if word == end:
                break
            sentence.append(word)
        return sentence

    def getSentenceProbability(self, sen):
        prob = 1.0
        lambda_val = 0.5  
        for i in range(len(sen) - 1):
            prev_word = sen[i]
            word = sen[i+1]
            bigram_prob = (self.bigram_counts[prev_word][word] + 1) / (self.unigram_counts[prev_word] + len(self.bigram_counts[prev_word]))
            unigram_prob = self.unigram_dist.prob(word)
            prob *= (lambda_val * bigram_prob) + ((1 - lambda_val) * unigram_prob)
        return prob

    def getCorpusPerplexity(self, corpus):
        log_prob_sum = 0.0
        total_words = 0
        for sen in corpus:
            total_words += len(sen) - 1  
            for i in range(len(sen) - 1):
                prev_word = sen[i]
                word = sen[i+1]
                bigram_prob = (self.bigram_counts[prev_word][word] + 1) / (self.unigram_counts[prev_word] + len(self.bigram_counts[prev_word]))
                unigram_prob = self.unigram_dist.prob(word)
                prob = (0.5 * bigram_prob) + (0.5 * unigram_prob)
                log_prob_sum += math.log(prob)
        avg_log_prob = log_prob_sum / total_words
        perplexity = math.exp(-avg_log_prob)
        return perplexity

# Sample class for a unsmoothed unigram probability distribution
class UnigramDist:
    def __init__(self, corpus):
        self.counts = defaultdict(float)
        self.total = 0.0
        self.train(corpus)

    def train(self, corpus):
        for sen in corpus:
            for word in sen:
                if word == start:
                    continue
                self.counts[word] += 1.0
                self.total += 1.0

    def prob(self, word):
        return self.counts[word] / self.total

    def draw(self):
        rand = random.random()
        for word in self.counts.keys():
            rand -= self.prob(word)
            if rand <= 0.0:
                return word

#-------------------------------------------
# The main routine
#-------------------------------------------
if __name__ == "__main__":
    trainCorpus = readFileToCorpus('train.txt')
    trainCorpus = preprocess(trainCorpus)
    
    posTestCorpus = readFileToCorpus('pos_test.txt')
    negTestCorpus = readFileToCorpus('neg_test.txt')
    
    vocab = set(word for sentence in trainCorpus for word in sentence)
    
    posTestCorpus = preprocessTest(vocab, posTestCorpus)
    negTestCorpus = preprocessTest(vocab, negTestCorpus)

    # Instantiate language models
    unigram_model = UnigramModel(trainCorpus)
    smoothed_unigram_model = SmoothedUnigramModel(trainCorpus)
    bigram_model = BigramModel(trainCorpus)
    smoothed_bigram_model = SmoothedBigramModelKN(trainCorpus)

    # Generate sentences and write to files
    unigram_model.generateSentencesToFile(20, "unigram_output.txt")
    smoothed_unigram_model.generateSentencesToFile(20, "smooth_unigram_output.txt")
    bigram_model.generateSentencesToFile(20, "bigram_output.txt")
    smoothed_bigram_model.generateSentencesToFile(20, "smooth_bigram_kn_output.txt")

    # Compute perplexity for test corpora
    pos_perplexity_unigram = unigram_model.getCorpusPerplexity(posTestCorpus)
    neg_perplexity_unigram = unigram_model.getCorpusPerplexity(negTestCorpus)

    pos_perplexity_smoothed_unigram = smoothed_unigram_model.getCorpusPerplexity(posTestCorpus)
    neg_perplexity_smoothed_unigram = smoothed_unigram_model.getCorpusPerplexity(negTestCorpus)

    pos_perplexity_bigram = bigram_model.getCorpusPerplexity(posTestCorpus)
    neg_perplexity_bigram = bigram_model.getCorpusPerplexity(negTestCorpus)

    pos_perplexity_smoothed_bigram = smoothed_bigram_model.getCorpusPerplexity(posTestCorpus)
    neg_perplexity_smoothed_bigram = smoothed_bigram_model.getCorpusPerplexity(negTestCorpus)

    print("Perplexity for positive test corpus (Unigram):", pos_perplexity_unigram)
    print("Perplexity for negative test corpus (Unigram):", neg_perplexity_unigram)

    print("Perplexity for positive test corpus (Smoothed Unigram):", pos_perplexity_smoothed_unigram)
    print("Perplexity for negative test corpus (Smoothed Unigram):", neg_perplexity_smoothed_unigram)

    print("Perplexity for positive test corpus (Bigram):", pos_perplexity_bigram)
    print("Perplexity for negative test corpus (Bigram):", neg_perplexity_bigram)

    print("Perplexity for positive test corpus (Smoothed Bigram):", pos_perplexity_smoothed_bigram)
    print("Perplexity for negative test corpus (Smoothed Bigram):", neg_perplexity_smoothed_bigram)
```
