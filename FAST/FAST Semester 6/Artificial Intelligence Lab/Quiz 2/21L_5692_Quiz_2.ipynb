```python
import random
```

New Population can be generated in a way that if a direct flight from Lahore to KArachi costs 1000 rupees but a flight from Lahore to Quetta costs 500 and from Quetta to Karachi costs 100 rupees the new population will help us in finding this route and hence propose this new cheaper way from Lahore -> Quetta -> Karachi  

```python
def mutation(parent1,parent2):

    # select the origin and destination at airports randomly as mutation points
    # by mutating new origin and destination the algorithm can make new population
    # the population will generate routes from new origins and destinations
    # the population can have a least price route from origin to destination
    mutation_point1 = random.randint(1, len(parent1) - 1)
    mutation_point2 = random.randint(mutation_point1, len(parent1))
    # Perform mutation to create two children
    child1 = parent1[:mutation_point1] + parent2[mutation_point1:mutation_point2] + parent1[mutation_point2:]
    child2 = parent2[:mutation_point1] + parent1[mutation_point1:mutation_point2] + parent2[mutation_point2:]
    return child1, child2
```

```python
# Generating fitness values for initial population

fitness_values = [fitnessfunction(chromosome) for chromosome in the population]
```

```python
# Generating new population by mutation

mutation_population_size = 10
new_generation = []

# Generating offspring through mutation
for _ in range(mutation_population_size):
    # Selecting parents based on fitness values
    parent1, parent2 = random.choices(population, k=2, weights=fitness_values)
    # Perform mutation to generate offspring
    child1, child2 = mutation(parent1, parent2)
    # Add offspring to new generation
    new_generation.extend([child1, child2])
```

```python
# Evaluate fitness of each itemset in the new generation

fitness_values_new_generation = [fitnessfunction(chromosome) for chromosome in new_generation]
```

```python
# Find the route with the highest fitness value or the least price
# max is being used as i am calculating the fitness valve not the price

best_fitness = max(fitness_values_new_generation)
```

```python
# Print the least price route

print("Best Route and Time and Price : ")
print(best_fitness)
```
