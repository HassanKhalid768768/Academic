```python
########### Question 1 ############
def symmetric_difference_count(set1, set2):
  # Your code here
  size1 = len(set1)
  size2 = len(set2)
  arr = []
  countset = 0

  for i in range(size1):
    temp = set1.pop()
    arr.append(temp)

  for i in range(size2):
    temp = set2.pop()
    arr.append(temp)

  for i in range(len(arr)):
    count = 0
    for j in range(len(arr)):
      if arr[i] == arr[j]:
        count = count + 1
    if count == 1:
      countset = countset + 1
  return countset

# Example usage:
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}
result = symmetric_difference_count(set1, set2)
print(result)
# Output should be the count of unique elements in the symmetric difference: 4
```

```python
############ Question 2 ##############
def is_disjoint_sets(set1, set2):

  # Your code here
  size1 = len(set1)
  size2 = len(set2)
  arr1 = []
  arr2 = []
  disjoint = 1

  for i in range(size1):
    temp = set1.pop()
    arr1.append(temp)

  for i in range(size2):
    temp = set2.pop()
    arr2.append(temp)

  for i in range(size1):
    temp = arr1[i]
    for j in range(size2):
      if temp == arr2[j]:
        disjoint = 0
        break

  if disjoint == 0:
    return ("Not Disjoint (Have elements in common)")
  else:
    return ("Disjoint (Have no elements in common)")

# Example usage:
set1 = {1, 2, 3, 4, 5}
set2 = {6, 7, 8, 9, 10}
result = is_disjoint_sets(set1, set2)
print(result)
# Output should be True because set1 and set2 have no elements in common.
```

```python
################ Question 3 #####################
def maximize_satisfaction(preferences):
    max_satisfaction = float('-inf')  # Initialize max satisfaction to negative infinity
    max_column = -1
    num_items = len(preferences[0])  # Get the number of items

    # Iterate over each item column-wise
    for i in range(num_items):
        total_satisfaction = 0

        # Calculate total satisfaction for current item
        for group in preferences:
            if i < len(group):
                total_satisfaction += list(group)[i]

        # Update max satisfaction and max column if necessary
        if total_satisfaction > max_satisfaction:
            max_satisfaction = total_satisfaction
            max_column = i + 1

    return max_satisfaction, max_column

# Example usage:
group1 = frozenset({1, 2, 3, 4, 5})
group2 = frozenset({3, 4, 5, 6, 7})
group3 = frozenset({1, 3, 5, 7, 9})
group4 = frozenset({2, 4, 6, 8, 10})

preferences = [group1, group2, group3, group4]
result_sum, result_column = maximize_satisfaction(preferences)
print("Max satisfaction sum:", result_sum)
print("Column number with max satisfaction:", result_column)
```

```python
################ Question 4 #####################
def has_cycle(graph):
    visited = set()

    def dfs(node, parent):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                if dfs(neighbor, node):
                    return True
            elif neighbor != parent:
                return True
        return False

    for node in graph:
        if node not in visited:
            if dfs(node, None):
                return True
    return False

# Example usage:
undirected_graph = {
    1: {2, 3},
    2: {1, 4},
    3: {1, 4},
    4: {2, 3}
}
result = has_cycle(undirected_graph)
print(result)

undirected_graph = {
    1: {2, 3},
    2: {1, 4},
    3: {1, 5},
    4: {2},
    5: {3}
}
result = has_cycle(undirected_graph)
print(result)
```

```python
################ Question 5 ###############
def calculate_average(numbers):
  # Your code here
    error = 1
    if not numbers:
        return ("List is empty. Cannot calculate the average.")
        error = 0
    for num in numbers:
        if not isinstance(num, (int, float)):
          return ("Invalid element type. All elements must be numerical.")
          error = 0

    if error != 0:
      return sum(numbers) / len(numbers)

# Example usage:
print("First Example : ")
thislist = [1, 2, 3, 4, 5, "banana"]
result = calculate_average(thislist)
print(result)
print()

print("Second Example : ")
thislist = []
result = calculate_average(thislist)
print(result)
print()

print("Third Example : ")
thislist = [1, 2.5, 3, 4, 5, 6.0]
result = calculate_average(thislist)
print(result)
```
