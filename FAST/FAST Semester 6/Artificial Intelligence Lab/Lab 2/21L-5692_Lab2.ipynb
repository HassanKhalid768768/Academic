```python
# Task 1
def removeValue(sampleList, val):
    return [x for x in sampleList if x != val]
```

```python
# Task 2
def findAndReplaceValue(sampleList, val):
    # get the first occurrence index
    index = sampleList.index(val)
    
    # update item present at location
    sampleList[index] = 200

    return sampleList
```

```python
# Task 3
def vehicleWeight(vehicleDict):
    return [key.upper() for key, value in vehicleDict.items() if value < 5000]
```

```python
# Task 4
def sortTuple(tupleSort):
    return tuple(sorted(tupleSort, key=lambda x: x[1]))
```

```python
# Task 5
def checkSameTuple(checkTuple):
    return all(x == checkTuple[0] for x in checkTuple)
```

```python
# Task 6
def stockTuple():
    return [
        ("25 Jan 2001", 43.50, 25, "CAT", 92.45),
        ("25 Jan 2001", 42.80, 50, "DD", 51.19),
        ("25 Jan 2001", 42.10, 75, "EK", 34.87),
        ("25 Jan 2001", 37.58, 100, "GM", 37.58)
    ]

def totalPurchasePrice(stocks):
    return sum(stock[1] * stock[2] for stock in stocks)

def totalProfitLoss(stocks):
    return sum((stock[4] * stock[2]) - (stock[1] * stock[2]) for stock in stocks)
```

```python
# Task 7
def deleteDict(sampleDict, keysToRemove):
    return {key: value for key, value in sampleDict.items() if key not in keysToRemove}
```

```python
# Task 8
def initializeDict(employees, defaults):
    return {employee: defaults.copy() for employee in employees}
```

```python
def main():

  list1 = [5, 20, 15, 20, 25, 50, 20]

  resList = removeValue(list1, 20)
  print(resList)

  resList = findAndReplaceValue(list1, 20)
  print(resList)

  vehicleDict = {"Sedan": 1500, "SUV": 2000, "Pickup": 2500, "Minivan": 1600, "Van": 2400, "Semi": 13600, "Bicycle": 7, "Motorcycle": 110}
  lst = vehicleWeight(vehicleDict)
  print(lst)

  tuple1 = (('a', 23),('b', 37),('c', 11), ('d',29))
  tuple1 = sortTuple(tuple1)
  print(tuple1)

  tuple1 = (45, 45, 45, 45)
  checkSame = checkSameTuple(tuple1)
  print(checkSame)

  print(totalPurchasePrice(stockTuple()))
  print(totalProfitLoss(stockTuple()))

  sampleDict = {
  "name": "Kelly",
  "age":25,
  "salary": 8000,
  "city": "New york"
  }
  keysToRemove = ["name", "salary"]
  print(deleteDict(sampleDict, keysToRemove))

  employees = ['Kelly', 'Emma', 'John']
  defaults = {"designation": 'Application Developer', "salary": 8000}

  resDict = initializeDict(employees, defaults)
  print(resDict)
```
