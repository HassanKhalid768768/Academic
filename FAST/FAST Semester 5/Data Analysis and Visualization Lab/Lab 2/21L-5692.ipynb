```python
import selenium
from selenium.webdriver.common.by import By
from selenium import webdriver
import csv
```

```python
driver = webdriver.Chrome()
url = 'https://www.youtube.com/@VillageFoodSecrets/videos'
driver.get(url)
```

```python
titles = driver.find_elements(By.XPATH, './/*[@id="video-title"]')
views = driver.find_elements(By.XPATH, './/*[@id="metadata-line"]/span[1]')
posted_times = driver.find_elements(By.XPATH, './/*[@id="metadata-line"]/span[2]')
```

```python
with open('youtube_videos.csv', 'w', newline='', encoding='utf-8') as csvfile:
    csv_writer = csv.writer(csvfile)
    
    csv_writer.writerow(["titles", "views", "posted time ago"])
    
    for title, view, posted_time in zip(titles, views, posted_times):
        csv_writer.writerow([title.text, view.text, posted_time.text])
```

```python
driver.quit()
```
