```python
def logistic(z: np.ndarray, a: float, b: float):
    log = 1 / ( 1 + ( b * np.exp(-a * z)))
    return log

log_ = logistic(my__array2d , -10, 10)
print("Answer : " , log_)
```

```python
def hyperbolic(z: np.ndarray, a: float, b: float, c: float):
    hyp = ( b * np.tanh( a * z )) + c
    return hyp

hyp_ = hyperbolic(my__array2d , 10 , -10 ,5)
print("Answer : " , hyp_)
```

```python
import sys
eps = sys.float_info.epsilon

def rectifiedlinearunit(z: np.ndarray):
    for i in range(len(z)):
        if z[i] > 0 :
            return z
        else:
            return z * eps 
    
rlu_ = rectifiedlinearunit(z=[3,16,16])
print("Answer : " , rlu_)
```

```python
from matplotlib import pyplot as plt
fig, ax = plt.subplots()
plt.title("Activation Functions")
plt.xlabel("x")
plt.ylabel("h(x)")
ax.plot(log_)
ax.plot(hyp_)
ax.plot(rlu_)
plt.show()
```
