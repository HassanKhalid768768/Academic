# Combined Code Version

```python
import pandas as pd
import plotly.express as px

# Task 2: Read Buildings data-set into a pandas format.
df = pd.read_csv("bird-window-collision-death.csv")

# Task 2: Display data of deaths of birds with respect to buildings
fig = px.pie(df, values='Deaths', names='Bldg #', title="Death with respect to Buildings", color="Side", hole=0.5)

# Task 3b: Change Color with respect to Building Sides
fig.update_traces(insidetextfont_color="white", textinfo="label+percent")

# Task 3c: Disable Clickable functionality of a Graph
fig.update_layout(legend_itemclick=False)

# Task 4b: Mention Building Number and Percentage on Pie Chart along with on Hover
fig.update_traces(hoverinfo="label+percent+name")

# Task 4a: Update Color of text inside the graph
fig.update_traces(insidetextfont_color="white")

# Task 1: Show the plot
fig.show()

# Task 1: Save the plot as an HTML file
fig.write_html("final_pie_chart.html")
```
