Task 1: Convert RGB image to grayscale

```python
import cv2
```

```python
my_image = cv2.imread('test_image.png')
```

```python
gray_image = (0.3 * my_image[:, :, 2] + 0.59 * my_image[:, :, 1] + 0.11 * my_image[:, :, 0]).astype('uint8')
```

```python
cv2.imshow('Grayscale Image', gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Task 2: Save data into JSON format

```python
import json
```

```python
gray_image_array = gray_image.tolist()
```

```python
json_string = json.dumps(gray_image_array)
```

```python
with open('grayimage.json', 'w') as f:
    f.write(json_string)
```

Task 3: Make a histogram and plot it

```python
import matplotlib.pyplot as plt
```

```python
my_image = cv2.imread('einstein1.webp')
histogram = cv2.calcHist([my_image], [0], None, [256], [0, 256])
```

```python
plt.hist(my_image.ravel(), bins=256, range=[0, 256])
plt.xlabel('Pixel Value')
plt.ylabel('Count')
plt.title('Histogram of Grayscale Image')
plt.show()
```

Task 4: Histogram normalization/equalization

```python
my_image = cv2.imread('einstein1.webp', cv2.IMREAD_GRAYSCALE)
equalized_image = cv2.equalizeHist(my_image)

plt.subplot(121), plt.imshow(my_image, cmap='gray')
plt.title('Original Image')
plt.subplot(122), plt.imshow(equalized_image, cmap='gray')
plt.title('Equalized Image')
plt.show()
```

```python
plt.hist(equalized_image.ravel(), bins=256, range=[0, 256])
plt.xlabel('Pixel Value')
plt.ylabel('Count')
plt.title('Histogram of Grayscale Image')
plt.show()
```

Task 5: Spatial filtering with a 3x3 kernel and resizing

```python
my_image = cv2.imread('test_image.png')
```

```python
resized_image = cv2.resize(my_image, (16, 16))
```

```python
import numpy as np 

kernel = np.array([[3, 0, 3],
                   [1, 0, 1],
                   [3, 0, 3]], dtype=np.float32)
```

```python
filtered_image = cv2.filter2D(resized_image, -1, kernel)
```

```python
cv2.imshow('Resized Image', resized_image)
cv2.imshow('Filtered Image', filtered_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
