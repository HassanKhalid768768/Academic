Declaring Variables

```r
# numeric
x <- 10.5
class(x)

# integer
x <- 1000L
class(x)

# complex
x <- 9i + 3
class(x)

# character/string
x <- "R is exciting"
class(x)

# logical/boolean
x <- TRUE
class(x)
```

Typecasting

```r
x <- 123
x_char <- as.character(x)
class(x)
class(x_char)
```

```r
x <- "123"
x_numeric <- as.numeric(x)
class(x)
class(x_numeric)
```

```r
x <- 123.45
x_integer <- as.integer(x)
class(x)
class(x_integer)
```

```r
x <- c(1, 0, 1, 1, 0)
x_logical <- as.logical(x)
class(x)
class(x_logical)
```

If-else Conditions

```r
x <- 10

if (x > 5) {
  result <- "x is greater than 5"
} else {
  result <- "x is not greater than 5"
}

cat("Result:", result, "\n")
```

Task 1

```r
myPower <- function(b, e) {
  result <- 1
  for (i in 1:e) {
    result <- result * b
  }
  return(result)
}

result1 <- myPower(2, 3)
cat("2^3 =", result1, "\n")
```

Task 2

```r
subtotal <- function(arr) {
  sum <- 0

  for (i in 1:length(arr)) {
    sum <- sum + arr[i]
    arr[i] <- sum
  }

  return(arr)
}

arr <- c(5.8, 2.6, 9.1, 3.4, 7.0)
modified_arr <- subtotal(arr)
average <- mean(arr)
print(modified_arr)
print(average)
```

```r
Counter <- function(char_array) {
  total_characters <- length(char_array)
  uppercase_letters <- sum(char_array %in% LETTERS)
  lowercase_letters <- sum(char_array %in% letters)
  spaces <- sum(char_array == " ")

  cat("No. of letters are:", total_characters - spaces, "\n")
  cat("No. of spaces are:", spaces, "\n")
  cat("No. of uppercase letters are:", uppercase_letters, "\n")
  cat("No. of lowercase letters are:", lowercase_letters, "\n")
}

input_string <- "Hello, I am a student of FAST University."
char_array <- strsplit(input_string, NULL)[[1]]
Counter(char_array)
```

```r
findMedianSortedArrays <- function(nums1, nums2) {
  merged <- sort(c(nums1, nums2))
  n <- length(merged)

  if (n %% 2 == 0) {
    return(mean(merged[c(n / 2, (n / 2) + 1)]))
  } else {
    return(merged[(n + 1) / 2])
  }
}

nums1 <- c(1, 2)
nums2 <- c(3, 4)
result <- findMedianSortedArrays(nums1, nums2)
cat("Output:", result, "\n")
```
