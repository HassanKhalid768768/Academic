```python
import dash
import dash_core_components as dcc
import dash_html_components as html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset (using Iris dataset as an example)
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# Initialize Dash app
app = dash.Dash(__name__)

# Define app layout with Graph components and Dropdown for chart type
app.layout = html.Div([
    html.H1("Interactive Dashboard"),

    # Dropdown for x-axis variable
    dcc.Dropdown(
        id='x-axis-dropdown',
        options=[
            {'label': col, 'value': col} for col in iris_df.columns[:-1]
        ],
        value=iris_df.columns[0],
        multi=False,
        style={'width': '50%'}
    ),

    # Dropdown for y-axis variable
    dcc.Dropdown(
        id='y-axis-dropdown',
        options=[
            {'label': col, 'value': col} for col in iris_df.columns[:-1]
        ],
        value=iris_df.columns[1],
        multi=False,
        style={'width': '50%'}
    ),

    # Dropdown for chart type
    dcc.Dropdown(
        id='chart-type-dropdown',
        options=[
            {'label': 'Scatter Plot', 'value': 'scatter'},
            {'label': 'Bar Plot', 'value': 'bar'},
            {'label': 'Box Plot', 'value': 'box'},
            {'label': 'Violin Plot', 'value': 'violin'},
            {'label': 'Histogram', 'value': 'histogram'},
            {'label': 'Line Plot', 'value': 'line'},
            {'label': 'Pie Chart', 'value': 'pie'},
            {'label': 'Heatmap', 'value': 'heatmap'},
        ],
        value='scatter',
        multi=False,
        style={'width': '50%'}
    ),

    # Graph component for the selected plot
    dcc.Graph(id='selected-plot')
])

# Define callback to update the selected plot based on dropdown selections
@app.callback(
    Output('selected-plot', 'figure'),
    [Input('x-axis-dropdown', 'value'),
     Input('y-axis-dropdown', 'value'),
     Input('chart-type-dropdown', 'value')]
)
def update_selected_plot(selected_x, selected_y, chart_type):
    if chart_type == 'scatter':
        fig = px.scatter(
            iris_df,
            x=selected_x,
            y=selected_y,
            color='target',
            labels={'target': 'Species'},
            title=f'Scatter Plot: {selected_x} vs {selected_y}'
        )
    elif chart_type == 'bar':
        fig = px.bar(
            iris_df,
            x='target',
            y=selected_x,
            labels={'target': 'Species', selected_x: selected_x},
            title=f'Bar Plot: {selected_x} by Species'
        )
    elif chart_type == 'box':
        fig = px.box(
            iris_df,
            x='target',
            y=selected_x,
            labels={'target': 'Species', selected_x: selected_x},
            title=f'Box Plot: {selected_x} by Species'
        )
    elif chart_type == 'violin':
        fig = px.violin(
            iris_df,
            x='target',
            y=selected_x,
            labels={'target': 'Species', selected_x: selected_x},
            title=f'Violin Plot: {selected_x} by Species'
        )
    elif chart_type == 'histogram':
        fig = px.histogram(
            iris_df,
            x=selected_x,
            color='target',
            labels={'target': 'Species'},
            title=f'Histogram: {selected_x} by Species'
        )
    elif chart_type == 'line':
        fig = px.line(
            iris_df,
            x='target',
            y=selected_x,
            labels={'target': 'Species', selected_x: selected_x},
            title=f'Line Plot: {selected_x} by Species'
        )
    elif chart_type == 'pie':
        fig = px.pie(
            iris_df,
            names='target',
            title=f'Pie Chart: {selected_x} by Species'
        )
    elif chart_type == 'heatmap':
        fig = px.imshow(
            iris_df.corr(),
            labels={'color': 'Correlation'},
            title=f'Heatmap: Correlation Matrix'
        )
    else:
        fig = px.scatter()  # Default to scatter plot if the chart type is not recognized

    return fig

# Run the app
if __name__ == '__main__':
    app.run_server(debug=True)
```
