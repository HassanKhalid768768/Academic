# TASK 6

```python
num = int(input("Enter a number for factorial calculation : "))    
factorial = 1    

if num < 0:    
    print(" Factorial does not exist for negative numbers")    
elif num == 0:    
    print("The factorial of 0 is 1")    
else:    
    for i in range(1, num+1):    
        factorial = factorial*i    
        
    print("Answer : ", factorial)    
```

```python
string1="A man, a plan, a canal: Panama"
string1 = string1.lower()
string2=[]
string3=[]
```

```python
def isitapalindrome(string1):
    size = len(string1)
    for l in range(size):
        if string1[l].isalnum():
            string2.append(string1[l])
    return string2
```

```python
isitapalindrome(string1)

string3 = string2[::-1]

if(string3==string2):
      print("The string is a palindrome")
else:
      print("Not a palindrome")
```

# Task 9.3

```python
def Sqrt(num):
 
    if (num == 0 or num == 1):
        return num
    
    i = 1
    result = 1
    while (result <= num):
 
        i += 1
        result = i * i
 
    return i - 1
```

```python
number = int(input("Input the number whose square root you want to calculate : "))
print("The Square root of the number is : ", Sqrt(number))
```
