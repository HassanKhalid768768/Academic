```python
from mrjob.job import MRJob

class countof_usernames(MRJob):
    
    def mapper(self, _, line):
        parts=line.split(',')
        UserName=parts[0]
        Comment=parts[1]
        yield(UserName,1)
        
    def reducer(self, UserName, count):
        yield(UserName,sum(count))
        
if __name__ = "__main__":
    countof_usernames.run()
        
```

# PART C

```python
from mrjob.job import MRJob

class maximumnumber_comments(MRJob):
    
    def mapper(self, _, line):
        parts=line.split(',')
        UserName=parts[0]
        Comment=parts[1]
        yield(UserName,(Comment,1))
        
    def reducer(self, UserName, count):
        yield(UserName,sum(count).max())
        
if __name__ = "__main__":
    maximumnumber_comments.run()
```

```python
from sklearn import train_test_split
data = train_test_split(train_size=70 , test_size=30)
```

```python
X = data['Quantity']
Y = data['UnitPrice']
data=list(zip(X,Y))
```

```python
inertias= []

from sklearn.cluster import k_means
kmeans=k_means(n_clusters=3)
kmeans.fit(data)

for i in range (3,6):
    kmeans=k_means(n_clusters=i)
    kmeans.fit(data)
    inertias.append(kmeans.inertias)
    
plt.scatter(range(3,6),inertias,c=kmeans.labels_)
plt.xlabel("Range")
plt.ylabel("Inertias")
plt.title("Elbow Method")
plt.plot()
    
```
