# Removing the outliers in Data_value

```python
# IQR
Q1 = np.percentile(data['Data_value'], 25,interpolation = 'midpoint')
 
Q3 = np.percentile(data['Data_value'], 75,interpolation = 'midpoint')

IQR = Q3 - Q1
 
print("Old Shape: ", data.shape)
 
# Upper bound
upper = np.where(data['Data_value'] >= (Q3+1.5*IQR))
# Lower bound
lower = np.where(data['Data_value'] <= (Q1-1.5*IQR))
 
''' Removing the Outliers '''
data.drop(upper[0], inplace = True)
data.drop(lower[0], inplace = True)
 
print("New Shape: ", data.shape)
```

# Removing the outliers in Period

```python
# IQR
Q1 = np.percentile(data['Period'], 25,interpolation = 'midpoint')
 
Q3 = np.percentile(data['Period'], 75,interpolation = 'midpoint')

IQR = Q3 - Q1
 
print("Old Shape: ", data.shape)
 
# Upper bound
upper = np.where(data['Period'] >= (Q3+1.5*IQR))
# Lower bound
lower = np.where(data['Period'] <= (Q1-1.5*IQR))
 
''' Removing the Outliers '''
data.drop(upper[0], inplace = True)
data.drop(lower[0], inplace = True)
 
print("New Shape: ", data.shape)
```

# Normalizing the period column of the dataset

```python
data['Period'] = data['Period'] /data['Period'].abs().max()
data
```

# store the feature matrix (X) and response vector (y)

```python
x = np.array(data['Age'])
y = np.array(data['Gpa'])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y,test_size=0.4, random_state=1)
```

```python
from sklearn import preprocessing
from sklearn import utils

#convert y values to categorical values
lab = preprocessing.LabelEncoder()
y_train = lab.fit_transform(y_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train.reshape(-1,1), y_train.reshape(-1,1).ravel())
```

```python
# making predictions on the testing set
y_pred = knn.predict(X_test.reshape(-1,1))

from sklearn import metrics

print("kNN model accuracy:", metrics.accuracy_score(y_test,y_pred))
```
